### Operando con datos numéricos

En Python, los datos numéricos (`int` y `float`) soportan diversas operaciones aritméticas:

- Suma (`+`)
- Resta (`-`)
- Multiplicación (`*`)
- División decimal (`/`)
- División entera (`//`)
- Módulo (`%`) (resto de la división)
- Potencia (`**`)

Veamos ejemplos prácticos:

````python
# Definimos algunos números
num1 = 5       # entero
num2 = 8       # entero
num3 = 9.5     # decimal (float)
````

````python
# Suma
suma = num1 + num3
print("Suma:", suma)
````

````python
# Resta
resta = num3 - num1
print("Resta:", resta)
````

````python
# Multiplicación
multi = num1 * num2
print("Multiplicación:", multi)
````

````python
# División decimal (resultado con decimales)
div_decimal = num2 / num1
print("División decimal:", div_decimal)
````

````python
# División entera (resultado sin decimales, se redondea hacia abajo)
div_entera = num2 // num1
print("División entera:", div_entera)
````

````python
# Módulo (resto de la división)
modulo = num2 % num1
print("Módulo:", modulo)
````

````python
# Potencia (num1 elevado a num2)
potencia = num1 ** 2
print("Potencia:", potencia)
````

### Operando con datos booleanos

Los valores booleanos (`True` y `False`) representan la lógica binaria. Permiten realizar operaciones lógicas y comparaciones:

- `and` → verdadero si ambos son verdaderos
- `or` → verdadero si al menos uno es verdadero
- `not` → invierte el valor lógico

También son el resultado de comparaciones, tales como: `<`, `<=`, `>`, `>=`, `==`, `!=`.

````python
# Booleanos básicos
a = True
b = False

# Operadores lógicos
print("a and b:", a and b)   # True si ambos son True
print("a or b:", a or b)     # True si al menos uno es True
print("not a:", not a)       # Invierte el valor lógico
````

````python
# Comparaciones numéricas que devuelven booleanos
x = 10
y = 3

print("x > y:", x > y)       # True
print("x < y:", x < y)       # False
print("x >= y:", x >= y)     # True
print("x <= y:", x <= y)     # False
print("x == y:", x == y)     # False
print("x != y:", x != y)     # True
````

### Operando con cadenas de texto

Las cadenas (`str`) o literales son secuencias de caracteres. Se pueden operar de varias formas:

- Concatenación (`+`)
- Repetición (`*`)
- Indexación (`[]`) y slicing (`[:]`)

Algunos métodos útiles son: `.upper()`, `.lower()`, `.title()`, `.replace()`, `.split()`

````python
# Definir cadenas
texto1 = "Hola"
texto2 = "Mundo"
````

````python
# Concatenación
saludo = texto1 + " " + texto2
print("Concatenación:", saludo)
````

````python
# Repetición
eco = texto1 * 3
print("Repetición:", eco)
````

````python
# Indexación y slicing
print("Primer carácter:", texto1[0])
print("Último carácter:", texto1[-1])
print("Subcadena usando slicing:", saludo[0:4])
````

````python
# Métodos útiles
print("Mayúsculas:", saludo.upper())
print("Minúsculas:", saludo.lower())
print("Título:", saludo.title())
print("Reemplazo:", saludo.replace("Mundo", "Python"))
print("Separación:", saludo.split(" "))
````

## Introducción a las funciones

En programación, una función es un bloque de código que realiza una tarea específica y que puede reutilizarse en diferentes partes de un programa. Son muy similares a las funciones matemáticas: reciben parámetros de entrada, realizan una transformación o cálculo, y devuelven un resultado.

En Python, para definir una función se utiliza la palabra clave `def`, seguida del nombre de la función, unos paréntesis (que pueden incluir parámetros de entrada) y un dos puntos (`:`) al final. El cuerpo de la función se escribe con indentación, y dentro se coloca la lógica. Para devolver un valor, se usa la palabra clave `return`.

A continuación, verás dos ejemplos de funciones: una para calcular el área de un triángulo y otra para calcular el área de un círculo.

````python
def calc_area_triangulo(base, altura):
  area = (base * altura) / 2
  return area
````

````python
area_t1 = calc_area_triangulo(2, 5)
print("Area triángulo 1:", area_t1)
````

````python
area_t2 = calc_area_triangulo(10, 3)
print("Area triángulo 2:", area_t2)
````
